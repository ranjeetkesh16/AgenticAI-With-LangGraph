```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict, Literal
from dotenv import load_dotenv
from pydantic import BaseModel, Field

load_dotenv()
```

```python
class SentimentSchema(BaseModel):
    sentiment: Literal["positive","negative"] = Field(desription="Sentiment of the review")
```

```python
class DiagnosisSchema(BaseModel):
    issue_type: Literal["UX/UI", "Performance", "Bug", "Support", "Other"] = Field(
        description="The category of issue"
    )
    tone: Literal["angry", "frustrated", "disappointed", "calm"] = Field(
        description="The emotional tone expressed"
    )
    urgency: Literal["Low", "Medium", "High"] = Field(
        description="How urgent or critical the issue appears to be"
    )
    
```
